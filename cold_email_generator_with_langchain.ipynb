```python
from langchain_community.document_loaders import WebBaseLoader
loader = WebBaseLoader("https://enterprise64.applytojob.com/apply/kzQMNKfGf1/ai-ml-engineer")
data = loader.load().pop().page_content
print(data)
```

```python
from langchain_core.prompts import PromptTemplate
prompt_extract = PromptTemplate.from_template(
        """
        ### SCRAPED TEXT FROM WEBSITE:
        {data}
        ### INSTRUCTION:
        The scraped text is from the career's page of a website.
        Your job is to extract the job postings and return them in JSON format containing the
        following keys: `role`, `experience`, `skills` and `description`.
        Only return the valid JSON.
        ### VALID JSON (NO PREAMBLE):
        """
)

chain_extract = prompt_extract | llm
res = chain_extract.invoke(input={'data':data})
print(res.content)
```

```python
from langchain_core.output_parsers import JsonOutputParser

json_parser = JsonOutputParser()
json_res = json_parser.parse(res.content)
json_res
```

```python
import pandas as pd
df=pd.read_csv("/content/my_portfolio.csv")
df
```

```python
import uuid
import chromadb

client = chromadb.PersistentClient('vectorstore')
collection = client.get_or_create_collection(name="portfolio")

if not collection.count():
    for _, row in df.iterrows():
        collection.add(documents=row["Techstack"],
                       metadatas={"links": row["Links"]},
                       ids=[str(uuid.uuid4())])
```

```python
links = collection.query(query_texts=job['skills'], n_results=2).get('metadatas', [])
links
```

```python
prompt_email = PromptTemplate.from_template(
        """
         ### JOB DESCRIPTION:
    {job_description}

    ### INSTRUCTION:
    You are an **AI/Machine Learning Engineer** specializing in building **scalable AI solutions, predictive models, and automation systems**.
    Your goal is to write a **concise, compelling cold email** demonstrating how your expertise in **AI, ML, and data-driven decision-making** can add value to the client’s business.

    **Key points to address:**
    - **Your AI/ML expertise:** Highlight your skills in machine learning, deep learning, and model deployment.
    - **Problem-solving approach:** Explain how you can leverage AI to solve business challenges such as **predictive analytics, automation, NLP, or computer vision** (whichever is relevant).
    - **Relevance to their job:** Connect your experience with the job description and explain how your technical skills align with their needs.
    - **Previous work:** Mention relevant projects or case studies (from {link_list}) to establish credibility.
    - **Call to action:** End with a clear next step, such as scheduling a discussion or exploring collaboration opportunities.

    The email should be **clear, professional, and impact-driven**, positioning you as a **skilled AI/ML engineer who can deliver innovative solutions**.

    Avoid generic introductions—focus on **value, relevance, and engagement**.

    ### EMAIL (NO PREAMBLE):
    """
)
chain_email = prompt_email | llm
res = chain_email.invoke({"job_description": str(job), "link_list": links})
print(res.content)
```
